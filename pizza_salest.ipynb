## KPI showing 

```python
# Total revenue
total_revenue = df['total_price'].sum()

# Total pizzas sold
total_pizzas_sold = df['quantity'].sum()

# Total unique orders
total_orders = df['order_id'].nunique()

# Average order value
avg_order_value = total_revenue / total_orders

# Average pizzas per order
avg_pizzas_per_order = total_pizzas_sold / total_orders

# Printing results with formatting
print(f"Total Revenue : {total_revenue:.2f}")
print(f"Total Pizzas Sold: {total_pizzas_sold}")
print(f"Total Orders: {total_orders}")
print(f"Avg Order Value : {avg_order_value:.2f}")
print(f"Average Pizza per Order : {avg_pizzas_per_order:.2f}")
```

```python
# Splitting ingredients, cleaning, counting, and summarizing
ingredient = (
    df['pizza_ingredients']       # select the column
        .str.split(',')           # split by comma into lists
        .explode()                # expand lists into separate rows
        .str.strip()              # remove extra spaces
        .value_counts()           # count frequency of each ingredient
        .reset_index()            # convert to DataFrame
        .rename(columns={'index':'Ingredients', 'pizza_ingredients':'count'})
)

# Display top 10 ingredients
print(ingredient.head(10))
```

## TREND ANALYSIS

```python
# Convert order_date column to datetime
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

# Extract day names
df['day_name'] = df['order_date'].dt.day_name()

# Define weekday order
weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

# Categorize days to maintain order
df['day_name'] = pd.Categorical(df['day_name'], categories=weekday_order, ordered=True)

# Group by day_name and count unique orders
orders_by_day = df.groupby('day_name', observed=False)['order_id'].nunique()

# Plot bar chart
ax = orders_by_day.plot(kind='bar', figsize=(8,5), color='red', edgecolor='black')

# Customize plot
plt.title('Total Orders by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)

# Add text labels on bars
for i, val in enumerate(orders_by_day):
    plt.text(i, val + 20, str(val), ha='center', va='bottom', fontsize=9, fontweight='bold')

# Adjust layout and show
plt.tight_layout()
plt.show()
```

```python
# Convert order_time to datetime (24-hour format)
df['order_time'] = pd.to_datetime(df['order_time'], format='%H:%M:%S')

# Extract hour from order_time
df['hour_of_day'] = df['order_time'].dt.hour

# Count distinct orders per hour
orders_per_hour = df.groupby('hour_of_day', observed=False)['order_id'].nunique()

# Plot bar chart with a different style
fig, ax = plt.subplots(figsize=(10, 5))
orders_per_hour.plot(kind='bar', ax=ax, color='darkblue', edgecolor='gray')

# Customize chart
ax.set_title('Orders Distribution by Hour')
ax.set_xlabel('Hour (0–23)')
ax.set_ylabel('Unique Orders')
ax.set_xticks(range(len(orders_per_hour)))
ax.set_xticklabels(orders_per_hour.index, rotation=0)

# Annotate bars with values
for idx, val in enumerate(orders_per_hour):
    ax.text(idx, val + 3, str(val), ha='center', va='bottom', fontsize=8, color='black')

plt.tight_layout()
plt.show()
```

```python
# Convert order_date to datetime
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

# Extract month names
df['month'] = df['order_date'].dt.month_name()

# Define month order
months = ["January", "February", "March", "April", "May", "June",
          "July", "August", "September", "October", "November", "December"]

# Categorize months to keep order
df['month'] = pd.Categorical(df['month'], categories=months, ordered=True)

# Count unique orders per month
orders_monthly = df.groupby('month', observed=False)['order_id'].nunique()

# Plot with a different style
plt.figure(figsize=(10, 5))
plt.bar(orders_monthly.index, orders_monthly.values, color="skyblue", edgecolor="black")

# Add line overlay for trend
plt.plot(orders_monthly.index, orders_monthly.values, color="red", marker='o', linewidth=2)

# Customize chart
plt.title("Monthly Orders Overview")
plt.xlabel("Month")
plt.ylabel("Orders Count")
plt.xticks(rotation=30)

# Annotate bars
for idx, val in enumerate(orders_monthly):
    plt.text(idx, val + 15, str(val), ha='center', fontsize=8, fontweight='bold')

plt.tight_layout()
plt.show()
```

## PERCENTAGE OF SALES/CATEGORY

```python
# Group sales by category
sales_by_category = df.groupby('pizza_category')['total_price'].sum()

# Calculate percentage contribution
sales_percentage = (sales_by_category / sales_by_category.sum()) * 100

# Plot with a different style
plt.figure(figsize=(8, 6))
explode = [0.05] * len(sales_percentage)   # explode all slices slightly
colors = plt.cm.Paired.colors              # use a different colormap

plt.pie(
    sales_percentage,
    labels=sales_percentage.index,
    autopct=lambda p: f'{p:.1f}%',
    startangle=120,
    colors=colors,
    explode=explode,
    shadow=True
)

plt.title("Pizza Category Sales Share")
plt.tight_layout()
plt.show()
```

## PERCENTAGE OF PIZZA SIZE/CATEGORY

```python
# Pivot table for sales by category and size
pivot_sales = df.pivot_table(
    index='pizza_category',
    columns='pizza_size',
    values='total_price',
    aggfunc='sum',
    fill_value=0
)

# Calculate percentage contribution
sales_share = (pivot_sales / pivot_sales.values.sum()) * 100

# Plot heatmap with different style
plt.figure(figsize=(9, 6))
sns.heatmap(
    sales_share,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=1,
    linecolor="white",
    cbar_kws={'label': 'Sales %'}
)

plt.title("Pizza Sales Share by Category & Size", fontsize=14, fontweight='bold')
plt.ylabel("Category")
plt.xlabel("Size")
plt.tight_layout()
plt.show()
```

## PIZZA SOLD/CATEGORY

```python
# Group pizzas sold by category
category_quantity = df.groupby('pizza_category')['quantity'].sum()

# Choose a different color palette
palette = plt.cm.Set3.colors[:len(category_quantity)]

# Plot using plt.bar instead of .plot()
plt.figure(figsize=(9, 5))
bars = plt.bar(category_quantity.index, category_quantity.values, 
               color=palette, edgecolor='gray')

# Customize chart
plt.title("Pizza Sales by Category", fontsize=14, fontweight='bold')
plt.xlabel("Category")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=30)

# Add labels above bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 10, str(height),
             ha='center', va='bottom', fontsize=8, color='black')

plt.tight_layout()
plt.show()
```

## TOP SELLING PIZZAS WITH THEIR TOTAL ORDERS

```python
# Group pizzas sold by name
pizza_sales = df.groupby('pizza_name')['quantity'].sum()

# Get top 5 pizzas
top_five = pizza_sales.sort_values(ascending=False).iloc[:5]

# Plot with a different style
plt.figure(figsize=(9, 5))
bars = plt.barh(top_five.index, top_five.values, color='teal', edgecolor='white')

# Customize chart
plt.title("Top Five Best-Selling Pizzas", fontsize=14, fontweight='bold')
plt.xlabel("Quantity Sold")
plt.ylabel("Pizza Name")

# Add labels on bars
for bar in bars:
    width = bar.get_width()
    plt.text(width + 5, bar.get_y() + bar.get_height()/2, str(width),
             ha='left', va='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

## SELLING PIZZAS WITH THEIR TOTAL ORDERS


```python
# Group orders by pizza name and count unique order IDs
pizza_order_counts = df.groupby('pizza_name')['order_id'].nunique()

# Get the top 5 pizzas
top_five = pizza_order_counts.sort_values(ascending=False).iloc[:5]

# Plot with a different style
plt.figure(figsize=(9, 5))
bars = plt.barh(top_five.index, top_five.values, 
                color='darkgreen', edgecolor='gray', alpha=0.9)

# Customize chart
plt.title("Top 5 Ordered Pizzas", fontsize=14, fontweight='bold')
plt.xlabel("Number of Orders")
plt.ylabel("Pizza Name")

# Add labels next to bars
for bar in bars:
    width = bar.get_width()
    plt.text(width + 2, bar.get_y() + bar.get_height()/2, str(width),
             ha='left', va='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Calculate revenue by pizza name
pizza_revenue = df.groupby('pizza_name')['total_price'].sum()

# Get top 5 pizzas by revenue
top_five_revenue = pizza_revenue.sort_values(ascending=False).iloc[:5]

# Plot with a different style
plt.figure(figsize=(9, 5))
bars = plt.barh(top_five_revenue.index, top_five_revenue.values, 
                color='crimson', edgecolor='gray', alpha=0.85)

# Customize chart
plt.title("Top 5 Revenue-Generating Pizzas", fontsize=14, fontweight='bold')
plt.xlabel("Revenue")
plt.ylabel("Pizza Name")

# Add labels beside bars
for bar in bars:
    width = bar.get_width()
    plt.text(width + 50, bar.get_y() + bar.get_height()/2, f"{width:.0f}",
             ha='left', va='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```
